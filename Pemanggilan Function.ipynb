Function abs()

```python
print("Suatu fungsi yang digunakan untuk menghitung jarak suatu nilai dari angka 0 dan")
print("jarak yang dihitung akan selalu bernilai positif. Nilai yanng didapat dari jarak")
print("tersebut biasa disebut dengan nilai absolute")
print("Contoh, nilai absolute dari (-5 - 3) adalah", abs(-5 - 3))
```
