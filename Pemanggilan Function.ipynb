```python
def kelilingLingkaran(r):
    return 2 * 3.14 * r
```

```python
def luasLingkaran(r):
    return 3.14 * r ** 2
```

```python
jarijari = float(input("Jari-jari lingkaran"))
keliling = kelilingLingkaran(jarijari)
luas = luasLingkaran(jarijari)
print("Lingkaran dengan jari-jari", str(round(jarijari))+"CM", "memiliki:")
print(f"Keliling = {keliling:.2f}CM")
print(f"Luas     = {luas:.2f}CM") 
```
