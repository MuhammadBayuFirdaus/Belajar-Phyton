Operator Perbandingan 

```python
print("Operator ==")
print("Membandingkan kedua value dengan cara 3 == 5 yang dapat diartikan apakah 3 sama ")
print("dengan 5 dengan hasil", 3 == 5, "bertipe data boolean")
```

```python
print("Operator !=")
print("Membandingkan kedua value dengan cara 3 != 5 yang dapat diartikan apakah 3 tidak")
print("sama dengan 5 dengan hasil", 3 != 5, "bertipe data boolean")
```

```python
print("Operator >")
print("Membandingkan kedua value dengan cara 3 > 5 yang dapat diartikan apakah 3 lebih")
print("besar dari pada 5 dengan hasil", 3 > 5, "bertipe data boolean")
```

```python
print("Operator <")
print("Membandingkan kedua value dengan cara 3 < 5 yang dapat diartikan apakah 3 lebih")
print("kecil dari pada 5 dengan hasil", 3 < 5, "bertipe data boolean")
```

```python
print("Operator >=")
print("Membandingkan kedua value dengan cara 3 >= 5 yang dapat diartikan apakah 3 lebih")
print("besar atau sama dengan 5 dengan hasil", 3 >= 5, "bertipe data boolean")
```

```python
print("Operator <=")
print("Membandinkan kedua value dengan cara 3 <= 5 yang dapat diartikan apakah 3 lebih")
print("kecil atau sama dengan 5 dengan hasil", 3 <= 5, "bertipe data boolean")
```

Operator Logika

```python
print("Operator and")
print("Membandingkan 2 kondisi jika kedua kondisi bernilai True maka akan mengembalikan")
print("value True dan jika setidaknya hanya satu kondisi yang bernilai False maka akan")
print("mengembalikan value False. Sebagai contoh:")
print("a. Jika 5 < 7 and 7 < 11 maka nilai yang akan dihasilkan adalah", 5 < 7 and 7 < 11)
print("b. Jika 5 < 7 and 7 > 11 maka nilai yang akan dihasilkan adalah", 5 < 7 and 7 > 11)
print("c. Jika 5 > 7 and 7 > 11 maka nilai yang akan dihasilkan adalah", 5 > 7 and 7 > 11)
```

```python
print("Operator or")
print("Membandingkan 2 kondisi jika kedua kondisi bernilai False maka akan mengembalikan")
print("value False dan setidaknya hanya satu kondisi yang bernilai True maka akan")
print("Mengembalikan value True. Sebagai contoh:")
print("a. Jika 5 < 7 or 7 < 11 maka nilai yang akan dihasilkan adalah", 5 < 7 or 7 < 11)
print("b. Jika 5 < 7 or 7 > 11 maka nilai yang akan dihasilkan adalah", 5 < 7 or 7 > 11)
print("c. Jika 5 > 7 or 7 > 11 maka nilai yang akan dihasilkan adalah", 5 > 7 or 7 > 11)
```

```python
print("Operator not")
print("Mengubah nilai suatu kondisi yang seharusnya bernial True akan diubah menjadi")
print("nilai False begitu juga sebaliknya. Sebagai contoh:")
print("Kondisi 5 < 7 yang harusnya adalah", 5 < 7, "jika diberi operator not akan mengembalikan")
print("nilai", not 5 < 7)
```

Operator identitas

```python
print("Operator is")
print("Terlihat dari luar operator ini bekerja hampir menyamai operator == namun jika")
print("operator == membandingkan kedua value operator is membandingkan kedua objek atau")
print("variabel. Sebagai contoh:")
a = 5
b = 5
print("a. Variabel a dan b yang sama-sama diisi dengan nilai 5 akan menampikan nilai", a is b)
print("   jika a dibandingkan dengan b")
b = 7
print("b. Variabel b yang diubah nilainya menjadi 7 akan menampilkan nilai", a is b)
print("   jika a dibandingkan dengan b")
```
