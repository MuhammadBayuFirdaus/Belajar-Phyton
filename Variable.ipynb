Variable Integer

```python
i = 7
print("Variabel i memiliki value", i, "dengan type data", type(i))
```

Variable Float

```python
f = 3.14
print("Variable f memiliki value", f, "dengan type data", type(f))
```

Variable String

```python
s = "Ayam"
print("Variable s memiliki value", s, "dengan type data", type(s))
```

Variable Boolean

```python
b = False
print("Variable b memiliki value", b, "dengan type data", type(b))
```

Variable List

```python
l = [1, 2, 3]
print("Variable l memiliki value", l, "dengan type data", type(l))
```

Variable Tuple

```python
t = (1, 2, 3)
print("Variable t memiliki value", t, "dengan type data", type(t))
```

Variable Set

```python
s = {1, 2, 3}
print("Variable s memiliki value", s, "dengan type data", type(s))
```

Variable Frozen Set

```python
fs = frozenset({1, 2, 3})
print("Variable fs memiliki value", fs, "dengan type data", type(fs))
```

Variable Dictionary

```python
d = {"Binatang" : "Ayam"}
print("Variable d memiliki value", d, "dengan type data", type(d))
```

Variable Range

```python
r = range(1, 3, 7)
print("Variable r memiliki value", r, "dengan type data", type(r))
```

Variable Byte

```python
bt = b"Ayam"
print("Variable bt memiliki value", bt, "dengan type data", type(bt))
```

Variable Byte Array

```python
bta = bytearray(b"Ayam")
print("Variable bta memiliki value", bta, "dengan type data", type(bta))
```

Variable Memory View

```python
mw = memoryview(b"Ayam")
print("Variable mw memiliki value", mw, "dengan type data")
print(type(mw))
```

Variable None

```python
n = None
print("Variable n memiliki value", n, "dengan ty[e data", type(n))
```
