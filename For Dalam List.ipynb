Perulangan berisi deret angka

```python
x = [i * 2 for i in range(10)]
print(x)
```

```python
x = [i % 3 == 0 for i in range(10)]
print(x)
```

```python
x = [i > 3  and i % 3 == 0 for i in range(10)]
print(x)
```

```python
a = "ayam"
b = [char for char in a]
print(b)
```

```python
a = "ayam"
b = [char+"a" for char in a]
print(b)
```

```python
a = "ayam"
b = [char.isalpha() for char in a]
print(b)
```
