Type data list

```python
print("Value", [1, 2], "dengan tyoe data", type([1, 2]))
print("Ini adalah type data yang berbentuk struktur data yang digunakan untuk menyimpan")
print("berbagai value")
print("Adapun karakteristik dari list adalah sebagai beikut:")
print("1. Isi dari list atau yang biasa disebut elemen memiliki urutan tertentu yang")
print("   dapat diakses dengan cara variable[0] dengan 0 sebagai index")
print("2. Elemen dalam list dapat diubah dengan cara sebagai berikut:")
print("   a. menggunakan .append(10) untuk memasukkan angka 10 kedakam variabel")
print("   b. menggunakan .remove(10) untuk menghapus angka 10 kedalam variabel")
print("3. Operator 'in' digunakan untuk mencari elemen terhadap value yang ingin")
print("   dicari dalam list dan mengembalikan type data boolean")
print("4. Menggunakan .index(10) untuk mecari angka 10 dalam list kemudian mengembalikan")
print("   index dari angka 10, jika tidak ditemukan akan mengembalikan ValueError")
print("5. Menggunakan .sort() yang akan mengurutkan semua elemen list secara ascending,")
print("   jika ingin mengurutkan secara descending maka gunakan .sort(reverse=True)")
print("6. Untuk mencari tau panjang dari sebuah list maka dapat menggunakan len(my_list)")
```

Type data set

```python
print("Ini value", {1, 2}, "dengan type data", type({1, 2}))
print("Type data yang hampir sama dengan list namun dengan berbagai ketentuan.")
print("Berikut merupakan uraian mengenai ketentuan type data set:")
print("1. Setiap element yang ada pada type data ini harus unik. Jika value yang akan")
print("   dimasukkan telah ada dalam variabel dengan type data ini maka value tersebut")
print("   tidak akan dimasukkan kedalam variabel")
print("2. Type data ini tidak memiliki urutan")
print("3. Pada dasarnya element pada set tidak bisa diubah secara langsung. Namun depat")
print("   diubah dengan cara sebagai berikut:")
print("   a. Menggabungkan 2 set dengan cara menggunakan operator | atau .union(). Sebagai")
print("      contoh, jika set pertama berisi {1, 2} dan set kedua berisi {2, 3} maka isi")
print("      set akan berubah menjadi {1, 2, 3}")
print("   b. Mengiris atau mencari persamaan antar 2 set dengan cara menggunakan operator")
print("      & atau .intersection(). Sebagai contoh, jika set pertama berisi {1, 2} dan")
print("      set kedua berisi {2, 3} maka isi set akan berubah menjadi {2}")
print("   c. Selisih atau mencari perbedaan antar 2 set dengan cara menggunakan operator -")
print("      atau .difference(). Sebagai contoh, jika set pertama berisi {1, 2} dan set")
print("      kedua berisi {2, 3} sedangkan yang ingin diubah adalah set pertama maka set")
print("      akan berubah menjadi {1}")
print("   d. .symmetric_difference() atau bisa menggunakan ^ memiliki fungsi yang hampir")
print("      sama dengan operator | namun menghilangkan set hasil operator &. Sebagai")
print("      contoh jika set pertama berisi {1, 2} dan set kedua berisi {2, 3} maka set")
print("      akan berubah menjadi {1, 3}")
```

Type data range

```python
print("Value", range(5), "dengan type data", type(range(5)))
print("Suatu type data yang dibuat dari object bernama range, type data ini sering")
print("digunakan dalam perulangan terutama untuk for")
print("Karena sebuah object, type ini memiliki beberapa parameter yang dapat diuraikaan")
print("sebagai berikut:")
print("1. range(stop)")
print("   Ketika digunakan fungsi diatas akan terlihat seperti ini range(3) maka ia akan")
print("   menghasilkan urutan 0,1,2")
print("2. range(start, stop)")
print("   Ketika digunakan fungsi diatas akan terlihat seperti ini range(3,5) maka ia akan" )
print("   menghasilkan urutan 3,4")
print("3. range(start, stop, step)")
print("   Ketika digunakan fungsi diatas akan terlihat seperti ini range(1,3,10) maka ia")
print("   akan menghasilkan urutan 1,4,7")
```

Type data bytearray

```python
print("Ini value", bytearray(b'Hello'), "dengan type data", type(bytearray(b'Hello')))
print("Type data ini sama dengan tipe data bytes namun dapat diubah layaknya mengubah")
print("list, sehingga beberapa fungsi seperti .append(), .remove() dan .index() dapat")
print("digunakan di type data ini")
print("Berikut fungsi yang hanya dapat digunakan dalam bytearray:")
print("1. .extend() digunakan untuk menyambung bytearray 1 dengan yang lain ataupun")
print("   dengan list dan tuple")
print("2. .insert(i, x) digunakan untuk mengubah value di index tertentu dengan x sebagai")
print("   value yang ingin dimasukkan dan i sebagai index")
print("3. .pop() berbeda dengan .remove() yang menggunakan element untuk menghapus array")
print("   .pop() menggunakan index untuk menghapus element. Secara default .pop() akan")
print("   menghapus element paling akhir")
print("4. .count() digunakan untuk menghitung seberapa banyak value yang dicari berada")
print("   dalam bytearray")
print("5. .reverse() membalik urutan array")
```
