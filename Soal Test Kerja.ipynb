Mencari angka yang hilang dalam suatu list

```python
def findNumber(number):
    lost = []
    max = 0
    for i in range(len(number)):
        if max < number[i]:
            max = number[i]

    for i in range(max):
        find = False
        for j in range(len(number)):
            if i == number[j]:
                find = True
        if find == False:
            lost.append(i)
    return lost
                
number = [5,3,7,0,8,6,1,10]
print(f'Angka yang ada {number}')
lostNumber = findNumber(number)
print(f'Angka yang hilang adalah {lostNumber}')
```

Melompati litersi menggunakan element yang ada pada list jika berhasi sampai ke elemen terakhir maka dinyatakan menang dan jika gagal dinyatakan kalah

```python
def jump(number):
    i = 0
    while i < len(number):
        i += number[i]

    if i == len(number):
        return 'Bisa'
    else:
        return 'Tidak bisa'

number = [1, 1, 3, 1, 1, 1]
atLast = jump(number)
print(f'{atLast} melompat sampai akhir')
```

Menghitung hutang yang harus dibayar dengan ketentuan tertentu

```python
def hutang(hari):
    if hari <= 3 :
        bayar = hari * 5000
    else:
        bayar = (3 * 5000) + ((hari - 3) * 10000)
        if bayar > 50000:
            bayar = bayar * 2
    return bayar

penundaan = 10
totalHutang = hutang(penundaan) 
print(f'Total hutang Rp.{totalHutang}')
```

Mencari element yang selalu berada di index 0 yang diartikan selalu menang

```python
def seleksi(pertandingan):
    menang = []
    kalah = []
    for i in range(len(pertandingan)):
        if pertandingan[i][0] not in menang:
            menang.append(pertandingan[i][0])
        if pertandingan[i][1] not in kalah:
            kalah.append(pertandingan[i][1])
    
    for i in range(len(kalah)):
        if kalah[i] in menang:
            menang.remove(kalah[i])

    hasil = [menang, kalah]
    return hasil

pertandingan = [[1,2],[3,4],[1,3],[10,9]]
juara = seleksi(pertandingan)
print(f'Pemenangnya adalah {juara[0]} dan yang kalah adalah {juara[1]}')
```

Menjumlahkan semua element yang ada pada list dan simpan setiap penjumlahanan didalam element

```python
def penjumlahan(baris):
    for i in range(len(baris)):
        if i > 0:
            baris[i] += baris[i - 1]
    return baris

baris = [1, 5, 9, 7, 5, 3]
print(f'Baris lama : {baris}')
hasil = penjumlahan(baris)
print(f'Baris baru : {hasil}')
```

Menghitung banyak literasi yang diperlukan untuk mengurangi value integer hingga menjadi 0

```python
def decreace(num, step, rekam):
    rekam.append(num)
    if num == 0:
        return step, rekam
    elif num % 2 == 0:
        return decreace(int(num / 2), step + 1, rekam)
    elif num % 2 != 0:
        return decreace(num - 1, step + 1, rekam)
    
number = 1024
step = 0
baris = []
hasil, rekam = decreace(number, step, baris)
print (f'{rekam}')
print(f'Banyak langkah yang dilalui {hasil}')
```

Make Diamond

```python
number = int(input())

for i in range(1, number + 1):
    level = ''
    if i < number:
        for j in range(number - i):
            level = level + ' '
    level = level + ('* ' * i)
    print(level)
    
for i in range(number - 1, 0, -1):
    level = ''
    if i < number:
        for j in range(number - i):
            level = level + ' '
    level = level + ('* ' * i)
    print(level)
```
