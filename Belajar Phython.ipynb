Value dan data type

```python
# Type data int atau integer
print("Value", 100, "dengan type data", type(100))
print("Ini adalah type data yang digunakan untuk value berupa bilangan bulat")
```

```python
# Type data float
print("Value", 3.14, "dengan type daya", type(3.14))
print("Ini adalah type data yang digunakan untuk value berupa bilangan desimal")
```

```python
# Type data complex
print("Value", 2j + 1, "dengan type data", type(2j + 1))
print("Ini adalah type data yang digunakan untuk value yang terdiri dari bilangan real")
print("dan imajiner")
```

```python
# Type data str atau string
print("Value", "Hello", "dengan type data", type("Hello"))
print("Ini adalah type data yang digunakan untuk value kalimat")
```

```python
# Type data bool atau boolean
print("Value", True, "dengan typr data", type(True))
print("Ini adalah type data yang digunakan untuk value yang berisi True ataupun False")
```

```python
# Type data list
print("Value", [1, 2], "dengan tyoe data", type([1, 2]))
print("Ini adalah type data yang berbentuk struktur data yang digunakan untuk menyimpan")
print("berbagai value")
print("Adapun karakteristik dari list adalah sebagai beikut :")
print("1. Isi dari list atau yang biasa disebut elemen memiliki urutan tertentu yang")
print("   dapat diakses dengan cara variable[0] dengan 0 sebagai index")
print("2. Elemen dalam list dapat diubah dengan cara sebagai berikut")
print("   a. menggunakan .append(10) untuk memasukkan angka 10 kedakam variabel")
print("   b. menggunakan .remove(10) untuk menghapus angka 10 kedalam variabel")
print("3. Menggunakan 'in' yang digunakan untuk mencari elemen terhadap value yang ingin")
print("   dicari dalam list dan mengembalikan type data boolean")
print("4. Menggunakan .index(10) untuk mecari angka 10 dalam list kemudian mengembalikan")
print("   index dari angka 10, jika tidak ditemukan akan mengembalikan ValueError")
print("5. Menggunakan .sort() yang akan mengurutkan semua elemen list secara ascending,")
print("   jika ingin mengurutkan secara descending maka gunakan .sort(reverse=True)")
print("6. Untuk mencari tau panjang dari sebuah list maka dapat menggunakan len(my_list)")
```

```python

print("Ini value", (1, 2), "dengan type data", type((1, 2)))
print("Ini value", range(5), "dengan type data", type(range(5)))
print("Ini value", {"nama" : "hasan"}, "dengan type data", type({"nama" : "hasan"}))
print("Ini value", {1, 2}, "dengan type data", type({1, 2}))
print("Ini value", frozenset([1, 2]), "dengan type data", type(frozenset([1, 2])))
print("Ini value", b'Hello', "dengan type data", type(b'Hello'))
print("Ini value", bytearray(b'Hello'), "dengan type data", type(bytearray(b'Hello')))
print("Ini value", memoryview(b"Hello"), "dengan type data", type(memoryview(b"Hello"))) 
print("Ini value", None, "dengan type data", type(None))
```
