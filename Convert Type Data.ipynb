Type Data String

```python
print("Tipe data String ke Integer")
print("7", "bertipe data", type("7"))
print("menjadi", int("7"), "bertipe data", type(int("7")))
```

```python
print("Tipe data String ke Float")
print("3.14", "bertipe data", type("3.14"))
print("menjadi", float("3.14"), "bertipe data", type(float("3.14")))
```

```python
print("Tipe data String ke Boolean")
print("ayam", "bertipe data", type("ayam"))
print("memjadi", bool("ayam"), "bertipe data", type(bool("ayam")))
```

```python
print("Tipe data String ke List")
print("ayam", "bertipe data", type("ayam"))
print("menjadi", list("ayam"), )
```
