```python
print("Tipe data Integer ke Float")
print(7, "bertipe data", type(7))
print("menjadi", float(7), "bertipe data", type(float(7)))
```

```python
print("Tipe data Integer ke Boolean")
print(7, "bertipe data", type(7))
print("menjadi", bool(7), "bertipe data", type(bool(7)))
```
