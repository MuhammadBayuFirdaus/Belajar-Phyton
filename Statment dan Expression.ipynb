Statment Integer

```python
i = (10 + 5) * 2
print("Expression dalam variable i adalah (10 + 5) * 2, namun value yang disimpan")
print("pada variabel adalah", i, "dengan type data", type(i))
```

Statment Float

```python
f = (10 + 5) / 3
print("Expression dalam variable f adalah (10 + 5) / 3, namun value yang disimpan")
print("pada variable adalah", f, "dengan type data", type(f))
```

Statment String

```python
s = "Ayam " + "Goreng"
print("Expression dalam variable s adalah 'Ayam ' + 'Goreng', namun value yang disimpan")
print("pada variable adalah", s, "dengan type data", type(s))
```

Statment Boolean

```python
b = True and False
print("Expression dalam variable b adalah True and False, namun value yang disimpan pada")
print("variable adalah", b, "dengan type data", type(b))
```

Statment List

```python
l = [1, 2, 3] + [3, 5, 7]
print("Expression dalam variable l adalah [1, 2, 3] + [3, 5, 7], namun value yang disimpan")
print("pada variable adalah", l, "dengan type data", type(l))
```

Statment Tuple

```python
t = (1, 2, 3) + (3, 5, 7)
print("Expression dalam variable t adalah (1, 2, 3) + (3, 5, 7), namun value yang disimpan")
print("pada variable adalah", t, "dengan type data", type(t))
```

Statment Set

```python
s = {1, 2, 3} ^ {3, 5, 7}
print("Expression dalam variable s adalah {1, 2, 3} ^ {3, 5, 7}, namun value yang disimpan")
print("pada variable adalah", s, "dengan type data", type(s))
```

Statment Frozen Set

```python
fs = frozenset({1, 2, 3}) ^ frozenset({3, 5, 7})
print("Expression dalam variable fs adalah frozenset({1, 2, 3}) ^ frozenset({3, 5, 7}),")
print("namun value yang dismpan pada variable adalah", fs, "dengan type")
print("data", type(fs))
```
