Golden Ratio

```python
import turtle

field = turtle.Screen()
x = turtle.Turtle()
i = 0
a = 1
b = 1
c = 0

while i <= 7 :
    c = a + b
    a = b
    b = c
    x.forward(c * 5)
    x.right(90)
    x.forward(c * 5)
    x.right(90)
    x.forward(c * 5)
    x.right(90)
    x.forward(c * 5)
    j = 0

    while j < 2:
        x.right(90)
        x.forward(c * 5)
        j += 1

    i += 1

turtle.bye()
turtle.done()
```
